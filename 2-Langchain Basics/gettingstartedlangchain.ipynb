```python
## Prompt Engineering

from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "You are an expert AI Engineer. Provide me answer based on the question."),
        ("user", "{input}")
    ]
)
```

```python
from langchain_groq import ChatGroq

llm3 = ChatGroq(model="gemma2-9b-it")
```

```python
## Chaining (Prompt ->> Model ->> Output)

chain = prompt|llm3
chain
```

```python
response = chain.invoke({"input": "Can you tell me something about LangSmith?"})
print(response)
```

```python
## OutputParser

from langchain_core.output_parsers import StrOutputParser

output_parser = StrOutputParser()

chain = prompt|llm3|output_parser
response = chain.invoke({"input": "Can you tell me something about LangSmith?"})
print(response)
```

```python
## JSONOutputParser

from langchain_core.output_parsers import JsonOutputParser

output_parser = JsonOutputParser()
```

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(
    template = "Answer the user query \n {format_instruction} \n {query}",
    input_variables = ["query"],
    partial_variables = {"format_instruction" : output_parser.get_format_instructions()}
)
```

```python
chain = prompt|llm3|output_parser
response = chain.invoke({"query": "Can you tell me something about LangSmith?"})
print(response)
```

```python
## Using Pydantic

from langchain_core.output_parsers import JsonOutputParser
from langchain_core.prompts import PromptTemplate
from langchain_openai import ChatOpenAI
from pydantic import BaseModel, Field

llm4 = ChatOpenAI(temperature=0)

class Joke(BaseModel):
    setup: str = Field(description = "question to setup joke.")
    punchline: str = Field(description = "Answer to resolve the joke.")

joke_query = "Tell me a joke on Payal Gaming"

output_parser = JsonOutputParser(pydantic_object = Joke)

prompt = PromptTemplate(
    template = "Answer the user query \n {format_instruction} \n {query}",
    input_variables = ["query"],
    partial_variables = {"format_instruction" : output_parser.get_format_instructions()}
)

chain = prompt|llm4|output_parser
chain.invoke({"query": f"{joke_query}"})
```
